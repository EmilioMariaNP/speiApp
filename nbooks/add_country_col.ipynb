<P>Adds a country column based on the NUTS2 value to a dataframe.<br>It is used to plot NUTS2 by country.</p>

```python
import pandas as pd
import pycountry
import numpy as np
```

```python
csv_file = '/home/politti/data/Danube_5min/spei_nuts_2_1min/copula_analysis/nuts_2_copula_analysis.csv'
```

```python
def get_country_name(code):
    try:
        # Convert code to uppercase to prevent matching failures
        country = pycountry.countries.get(alpha_2=str(code).upper())
        return country.name if country else "Unknown"
    except (AttributeError, ValueError):
        return "Unknown"
```

```python
df = pd.read_csv(csv_file)
df.head(1)
```

```python
df['CNT'] = df['NUTS'].str[:2]
df.head(1)
```

```python
df['Country'] = df['CNT'].apply(get_country_name)
df.head()
```

```python
# df[df['country'] == 'Unknown'][['NUTS', 'CNT']]

xk_mask = df['CNT'] == 'XK'
df['Country'] = np.where(xk_mask,'Kosovo', df['Country'])
```

```python
df['Country'] = df['Country'].replace({'Moldova, Republic of': 'Republic of Moldova'})
```

```python
print(sorted(df.country.unique().tolist()) )
```

```python
df.to_csv(csv_file, index = False)
```

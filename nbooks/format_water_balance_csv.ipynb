<p>Formats scenarios columns of water balance for better display in charts.</p>

```python
import pandas as pd
```

```python
sspsrcp_labels =  ['Historical', 'SSP1-2.6', 'SSP2-4.5', 'SSP3-7.0', 'SSP5-8.5']
sspsrcp_values =  ['historical', 'ssp126', 'ssp245', 'ssp370', 'ssp585']
value_labels_dict = dict(zip(sspsrcp_values, sspsrcp_labels))
```

```python
water_balance_csv_nuts = '/home/politti/data/Danube_5min/spei_nuts_2/water_balance_danube_nuts2.csv'
water_balance_csv_cru = '/home/politti/data/Danube_5min/spei_nuts_2/water_balance_danube_nuts2_cru.csv'
```

```python
df = pd.read_csv(water_balance_csv_nuts)
#df['scenario'].unique()

# rename ssp-rcp for better charts display
df['scenario'] = df.scenario.replace(value_labels_dict)
#df['scenario'].unique()

# rename shapefile-object id column
df.rename(columns={'OBJECTID': 'nuts2_id'}, inplace=True)
df['nuts2_id'] = df['nuts2_id'].astype(int)

df.to_csv(water_balance_csv_nuts, index=False)
df.head(3)
```

```python
df_cru = pd.read_csv(water_balance_csv_cru)
df_cru.rename(columns={'OBJECTID': 'nuts2_id'}, inplace=True)
df_cru['nuts2_id'] = df_cru['nuts2_id'].astype(int)
df_cru.to_csv(water_balance_csv_cru, index=False)
df_cru.head(3)
```

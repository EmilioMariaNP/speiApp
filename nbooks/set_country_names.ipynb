```python
import geopandas as gpd
import pycountry
import pandas as pd
```

```python
nuts_2_countries_shp = '/home/politti/data/Danube_baselayers/danube_nutw2_countries.shp'
gdf = gpd.read_file(nuts_2_countries_shp)
gdf.head()
```

```python
def get_country_info(iso2_code):
    try:
        # Look up the country object using the alpha_2 code
        country = pycountry.countries.get(alpha_2=iso2_code)
        # Return a tuple: (ISO3, Full Name)
        return country.alpha_3, country.name
    except (LookupError, AttributeError):
        # Return None for both if the code is invalid or missing
        return None, None
```

```python
# 3. Apply the function and unpack into your two new columns
gdf[['ISO3', 'Country']] = gdf['NUTS0'].apply(
    lambda x: pd.Series(get_country_info(x))
)
gdf.rename(columns={'NUTS0': 'ISO2'}, inplace=True)
gdf = gdf[['ISO2', 'ISO3', 'Country', 'geometry']]
gdf.head(18)
```

```python
gdf.to_file(nuts_2_countries_shp)
```

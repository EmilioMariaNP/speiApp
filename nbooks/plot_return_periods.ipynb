```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
import numpy as np

def set_context(fig_size = (21.7,21.27), contx = 'paper', axes_style = 'darkgrid'):
    sns.set_style("darkgrid")
    sns.set_context(contx)
    sns.set_theme(rc={'figure.figsize':fig_size}) #width - height
    sns.axes_style(axes_style)
    return 0

set_context(fig_size = (16,11), contx = 'talk')
```

```python
copula_csv = '/home/politti/data/Danube_5min/spei_nuts_2/copula_analysis/nuts_2_copula_analysis.csv'
spatial_unit_col = 'NUTS_ID'
return_periods = [5, 10, 30, 50]
```

```python
df = pd.read_csv(copula_csv)
df.head(1)
```

```python
# Filter the DataFrame for the desired SPEI index
df_filtered = df[(df['spei'] == 3) & (df['Return period'].isin(return_periods))]

# Create the FacetGrid
g = sns.FacetGrid(
    df_filtered,
    row="NUTS_ID",
    col="scenario",
    height=4,
    aspect=1.5,
    margin_titles=True
)

# Map a line plot onto the grid, using median as the estimator
g.map_dataframe(
    sns.pointplot,
    x="Return period",
    y="Likelihood_Change",
    estimator=np.median,
    errorbar='ci',
    linestyle='none',
    #join=False
)

# Set titles and labels for clarity
g.set_axis_labels("Return Period", "Likelihood Change (Median)")
g.set_titles(col_template="{col_name}", row_template="{row_name}")

# Adjust layout and display the plot
plt.tight_layout()
plt.show()
```
